```python
print("===== SMART TRAVEL AGENT =====")

destination = input("Enter destination (Murree/Lahore/Hunza): ").lower()
budget = int(input("Enter your budget: "))
trip_type = input("Enter trip type (family/adventure): ").lower()

if destination == "murree":
    if budget >= 30000 and trip_type == "family":
        print("Recommended: Murree Family Trip")
    else:
        print("Budget or trip type is not suitable for Murree.")

elif destination == "lahore":
    if budget >= 25000:
        print("Recommended: Lahore City Trip")
    else:
        print("Budget is low for Lahore.")

elif destination == "hunza":
    if budget >= 50000 and trip_type == "adventure":
        print("Recommended: Hunza Adventure Trip")
    else:
        print("Budget or trip type is not suitable for Hunza.")

else:
    print("Destination not available.")
```
